```python
# Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# Paths
RAW_DATA = Path("../data/generated")
PROCESSED_DATA = Path("../data/processed")
```

```python
# Load data
customers = pd.read_csv(RAW_DATA / "customers_data.csv")
sales = pd.read_csv(RAW_DATA / "sales_data.csv")
products = pd.read_csv(RAW_DATA / "products_data.csv")
marketing = pd.read_csv(RAW_DATA / "marketing_data.csv")
```

```python
# Display data
display(customers.head())
display(sales.head())
display(products.head())
display(marketing.head())
```

```python
# Shapes
print("Customers:", customers.shape)
print("Sales:", sales.shape)
print("Products:", products.shape)
print("Marketing:", marketing.shape)
```

```python
# Info
customers.info()
sales.info()
products.info()
marketing.info()
```

```python
# Missing values
print("Customers")
print(customers.isnull().sum())

print("\nSales")
print(sales.isnull().sum())

print("\nProducts")
print(products.isnull().sum())

print("\nMarketing")
print(marketing.isnull().sum())
```

```python
# Duplicates
print("Customers:", customers.duplicated().sum())
print("Sales:", sales.duplicated().sum())
print("Products:", products.duplicated().sum())
print("Marketing:", marketing.duplicated().sum())
```

```python
# Statistics
display(customers.describe(include="all"))
display(sales.describe(include="all"))
display(products.describe(include="all"))
display(marketing.describe(include="all"))
```

```python
# Convert dates
customers["Join_Date"] = pd.to_datetime(customers["Join_Date"])
sales["Date"] = pd.to_datetime(sales["Date"])
marketing["Start_Date"] = pd.to_datetime(marketing["Start_Date"])
marketing["End_Date"] = pd.to_datetime(marketing["End_Date"])
```

```python
# Check dates
print(customers.dtypes)
print(sales.dtypes)
print(marketing.dtypes)
```

```python
# Validate customers
print("Age min:", customers["Age"].min())
print("Age max:", customers["Age"].max())
print("Spent min:", customers["Total_Spent"].min())
```

```python
# Validate sales
print("Quantity min:", sales["Quantity"].min())
print("Sale price min:", sales["Sale_Price"].min())
print("Channels:", sales["Channel"].unique())
```

```python
# Validate products
print("Price min:", products["Price"].min())
print("Categories:", products["Category"].unique())
print("Brands:", products["Brand"].unique())
```

```python
# Validate marketing
print("Budget min:", marketing["Budget"].min())
print("Impressions min:", marketing["Impressions"].min())
print("Clicks min:", marketing["Clicks"].min())
print("Conversions min:", marketing["Conversions"].min())
```

```python
# Check customer references
print(
    sales["Customer_ID"].isin(customers["Customer_ID"])
)
```

```python
# Check product references
print(
    sales["Product_ID"].isin(products["Product_ID"])
)
```

```python
# Check campaign consistency
print(
    (marketing["Conversions"] <= marketing["Clicks"]).all()
)

print(
    (marketing["Clicks"] <= marketing["Impressions"]).all()
)
```

```python
# Check campaign dates
print(
    (marketing["Start_Date"] <= marketing["End_Date"]).all()
)
```

```python
# Calculate revenue
sales["Revenue"] = sales["Sale_Price"]

# Display revenue
display(sales)
```

```python
# Total revenue
print("Total revenue:", sales["Revenue"].sum())
```

```python
# Comparer les prix
sales_products = sales.merge(
    products[["Product_ID", "Price"]],
    on="Product_ID",
    how="left"
)

display(sales_products[[
    "Product_ID",
    "Quantity",
    "Sale_Price",
    "Price"
]])
```

```python
# Vérifier les prix
sales_products["Expected_Total"] = (
    sales_products["Quantity"] * sales_products["Price"]
)

sales_products["Price_Consistent"] = (
    sales_products["Sale_Price"] == sales_products["Expected_Total"]
)

display(sales_products)
print(
    "Prix conformes :",
    sales_products["Price_Consistent"].sum(),
    "/",
    len(sales_products)
)
```

```python
# Vérifier les écarts
sales_products["Expected_Total"] = (
    sales_products["Quantity"] * sales_products["Price"]
)

display(sales_products)
```

```python
# Jointure ventes-produits
sales_full = sales.merge(
    products,
    on="Product_ID",
    how="left"
)

# Jointure clients-ventes
sales_full = sales_full.merge(
    customers,
    on="Customer_ID",
    how="left"
)

display(sales_full)
```

```python
# Save cleaned data
customers.to_csv(
    PROCESSED_DATA / "customers_data.csv",
    index=False
)

sales.to_csv(
    PROCESSED_DATA / "sales_data.csv",
    index=False
)

products.to_csv(
    PROCESSED_DATA / "products_data.csv",
    index=False
)

marketing.to_csv(
    PROCESSED_DATA / "marketing_data.csv",
    index=False
)
```

```python
# Statistiques clients
customer_analytics = sales_full.groupby("Customer_ID").agg(
    Number_Of_Purchases=("Sale_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Sales_Revenue=("Revenue", "sum"),
    Average_Purchase=("Revenue", "mean")
).reset_index()

display(customer_analytics)
```

```python
# Informations clients
customer_analytics = customer_analytics.merge(
    customers[
        ["Customer_ID", "Age", "Gender", "Location", "Join_Date", "Total_Spent"]
    ],
    on="Customer_ID",
    how="left"
)

display(customer_analytics)
```

```python
# Sauvegarder
customer_analytics.to_csv(
    PROCESSED_DATA / "customer_analytics.csv",
    index=False
)
```
